```python
%matplotlib widget
"""Every instrument's figure behind one drop-down, for presenting: serve it alone with

    voila app.ipynb

Only the first instrument is built at the start; the others the first time they are
chosen, and then kept, so switching back finds them as they were left. The figures are
sized to fit, with their controls, a 1920 x 1080 slide.
"""
from IPython.display import display
from ipywidgets import Dropdown, Layout, VBox

from bifrost import BIFROST
from in5 import IN5
from phasespace import figure
from trex import TREX

#: what the drop-down offers, in order, each with its starting settings
INSTRUMENTS = {
    'BIFROST, ESS': (BIFROST, dict(energy_min=4.5, shaping_time=0.2)),
    'T-REX, ESS': (TREX, dict(wavelength=2.4, rrm=4, mode='HF')),
    'IN5, ILL': (IN5, dict(wavelength=4.5, speed=8500, ratio=0.5, counter=True)),
}
FIGSIZE = (12.5, 8.6)   # inches
built = {}


def show(name):
    if name not in built:
        instrument, initial = INSTRUMENTS[name]
        built[name] = figure(instrument, figsize=FIGSIZE, **initial)
    page.children = [choose, built[name]]


choose = Dropdown(options=list(INSTRUMENTS), description='instrument',
                  layout=Layout(width='20em'))
page = VBox()
choose.observe(lambda change: show(change['new']), names='value')
show(choose.value)
display(page)
```
